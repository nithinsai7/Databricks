### Data Objects - Tables and Views

```sql
create view acc_dev.nithin.sales_vw as 
select * from acc_dev.nithin.sales;
```

```sql
create temp view sales_vw as 
select product_id , sum(total_amount) as total from acc_dev.nithin.sales
group by product_id
order by total desc;
```

```sql
select * from sales_vw
```

## UDF


```sql
%sql
create or replace function acc_dev.nithin.full_name(firstname string, lastname string)
returns string
return concat(firstname + ' ' + lastname);
```

```sql
%sql
select acc_dev.nithin.full_name("nithin","reddy") as full_name
```
