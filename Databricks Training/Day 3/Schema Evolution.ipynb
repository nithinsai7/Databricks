```sql
use catalog acc_dev;
use schema nithin;
```

```sql
create table cricket (id int , name string);
```

```sql
%python
data = ([(1,"Virat") , (2,"Rohit")])
schema = "id int , name string"

df = spark.createDataFrame(data,schema)
df.display()
```

```sql
%python
df.write.mode("overwrite").saveAsTable("cricket")
```

```sql
%python
data = ([(3,"Sachin") , (4,"Bumrah")])
schema = "id int , name string"

df = spark.createDataFrame(data,schema)
df.display()
```

```sql
%python
data = ([(5,"Chahal" , "Spinner") , (6,"Bhuvaneshawar" , "Bowler")])
schema = "id int , name string , role string"

df = spark.createDataFrame(data,schema)
df.display()
```

```sql
%python
df.write.mode("append").option("mergeSchema", "true").saveAsTable("cricket")
```
