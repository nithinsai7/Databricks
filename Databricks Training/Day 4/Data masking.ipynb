```python
%sql
CREATE OR REPLACE FUNCTION datamask(x STRING)
  RETURNS STRING
  RETURN CONCAT(REPEAT("*", LENGTH(x) - 2), RIGHT(x, 2)
); 
```

```python
%sql
drop view if exists nithin_gold.agg_heartrate;

CREATE or replace VIEW nithin_gold.agg_heartrate AS
SELECT
  CASE WHEN
    is_account_group_member('da') THEN acc_dev.nithin_silver.datamask(mrn)
    ELSE mrn
  END AS mrn,
  time,
  device_id,
  heartrate
FROM heartrate_device
WHERE
  CASE WHEN
    is_account_group_member('account users') THEN device_id < 30
    ELSE TRUE
  END
```
